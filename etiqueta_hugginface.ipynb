### Cargar datos desde Google Sheets

Primero, necesitas obtener el enlace directo de descarga CSV para tu hoja de cálculo de Google Sheets. Puedes hacerlo yendo a `Archivo > Compartir > Publicar en la web` y luego eligiendo `CSV` para la hoja específica.

```python
import pandas as pd

# Reemplaza esta URL con el enlace directo de descarga CSV de tu Google Sheet.
# Asegúrate de que sea el enlace de publicación en la web como CSV.
csv_url = 'https://docs.google.com/spreadsheets/d/1jtQmo16nXKevwKXe-F-DkHPoBVyM9xpaeWAXx-L1WtA/export?format=csv'

try:
    df = pd.read_csv(csv_url)
    print("Datos cargados exitosamente.")
except Exception as e:
    print(f"Error al cargar el archivo CSV: {e}")
    df = pd.DataFrame() # Crear un DataFrame vacío en caso de error

# Mostrar las primeras filas del DataFrame original
if not df.empty:
    print("Primeras 5 filas del DataFrame:")
    display(df.head())
else:
    print("El DataFrame está vacío, no se pueden mostrar las primeras filas.")
```

### Crear la columna 'Tipo_requisito'

Ahora, crearemos la nueva columna 'Tipo_requisito' basándonos en los valores de la columna 'FINAL_LABEL'.

```python
if not df.empty and 'FINAL_LABEL' in df.columns:
    df['Tipo_requisito'] = df['FINAL_LABEL'].apply(lambda x: 'RNF' if x == 'NF' else ('RF' if x == 'F' else None))
    print("Columna 'Tipo_requisito' creada exitosamente.")
    # Mostrar las primeras filas del DataFrame con la nueva columna
    print("Primeras 5 filas del DataFrame con 'Tipo_requisito':")
    display(df.head())
elif df.empty:
    print("El DataFrame está vacío. No se pudo crear la columna 'Tipo_requisito'.")
else:
    print("La columna 'FINAL_LABEL' no se encontró en el DataFrame.")
```

### Guardar el DataFrame en un archivo CSV

Vamos a guardar el DataFrame modificado en un archivo CSV local para que puedas descargarlo.

```python
# Guarda el DataFrame en un archivo CSV
output_csv_filename = 'dataset_hugginface_spanish_etiqueta.csv'
df.to_csv(output_csv_filename, index=False)

print(f"El DataFrame se ha guardado exitosamente en '{output_csv_filename}'.")
print("Puedes descargar este archivo desde el panel de Archivos de Colab (icono de carpeta a la izquierda).")
```
